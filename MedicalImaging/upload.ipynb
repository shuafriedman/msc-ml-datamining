```python
import os
import shutil
from huggingface_hub import HfApi, HfFolder, Repository
```

```python

HF_TOKEN = input("Enter your Hugging Face token: ")
HF_USERNAME = "Shuaf98"
```

```python
MODEL_DIR = "/teamspace/studios/this_studio/test_results/models"
```

```python
# Loop through the folder and upload each model
for model_file in os.listdir(MODEL_DIR):
    if model_file.endswith(".pth"):
        model_name = model_file.replace(".pth", "")
        model_path = os.path.join(MODEL_DIR, model_file)

        # Create the repository if it doesn't exist
        repo_id = f"{HF_USERNAME}/{model_name}"
        api = HfApi()
        api.create_repo(repo_id=repo_id, token=HF_TOKEN, exist_ok=True)

        # Clone the repository locally using the full path (username/repo_name)
        repo_url = f"https://huggingface.co/{repo_id}"
        repo = Repository(local_dir=model_name, clone_from=repo_url, use_auth_token=HF_TOKEN)

        # Copy the .pth file to the repo folder and commit
        shutil.copy(model_path, os.path.join(model_name, model_file))

        # Push the model to the repository
        repo.git_add(auto_lfs_track=True)  # Tracks large files with Git LFS
        repo.git_commit("Add model.pth")
        repo.git_push()
        
        print(f"Uploaded {model_file} to Hugging Face as {model_name}")

print("Model upload complete!")
```
